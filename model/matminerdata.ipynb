```python
#df_mp= pd.read_csv("battery.csv")
import numpy as np
```

```python
from matminer.featurizers.composition import ElementProperty
from pymatgen.core import Composition
```

```python
from pymatgen.core.periodic_table import Element
dataset= Element
```

```python
df_mp["composition"] = df_mp['formula_pretty'].map(lambda x: Composition(x))
```

```python
df_mp['vpa'] = df_mp['volume']/df_mp['nsites']
```

```python
descriptors = ['row', 'group', 'atomic_mass',
               'atomic_radius', 'boiling_point', 'melting_point', 'X']
stats = ["mean", "std_dev"]

ep = ElementProperty(data_source=dataset, features=descriptors, stats=stats)
df_mp = ep.featurize_dataframe(df_mp, "composition",ignore_errors=True)
```

```python
df_mp.info()
```

```python
df_mp = df_mp.dropna()
```
