```python
import os
import pandas as pd
from ase.io import read  # ASE for reading CIF files
from pymatgen.io.ase import AseAtomsAdaptor  # For converting ASE to Pymatgen
from matminer.featurizers.composition import ElementProperty

# Directory containing CIF files
cif_dir = "felix"

# Initialize ElementProperty featurizer with the "magpie" preset
ep = ElementProperty.from_preset("magpie")

# List to hold composition names and their features
compositions = []
features = []

# Loop through each CIF file in the directory
for cif_file in os.listdir(cif_dir):
    if cif_file.endswith(".cif"):
        # Load structure from CIF using ASE
        ase_structure = read(os.path.join(cif_dir, cif_file))
        
        # Convert ASE Atoms to pymatgen Structure using AseAtomsAdaptor
        structure = AseAtomsAdaptor.get_structure(ase_structure)
        
        # Access the composition attribute directly from the structure
        composition = structure.composition
        
        # Featurize composition using ElementProperty
        composition_features = ep.featurize(composition)
        
        # Store the composition's formula and its features
        compositions.append(composition.reduced_formula)
        features.append(composition_features)

# Create a DataFrame with compositions and their features
feature_labels = ep.feature_labels()  # Get feature labels from ElementProperty
df = pd.DataFrame(features, columns=feature_labels)
df.insert(0, 'Composition', compositions)  # Add composition as the first column

# Save the DataFrame to a CSV file
df.to_csv('composition_features.csv', index=False)

print(f"Feature CSV file has been created with {len(df)} entries.")
```

```python
import os
from ase.io import write
from ase.build import bcc111

# List of transition metals (excluding La and Ac)
transition_metals = ['Sc', 'Ti', 'V', 'Cr', 'Mn', 'Fe', 'Co', 'Ni', 'Cu', 'Zn',
                     'Y', 'Zr', 'Nb', 'Mo', 'Ru', 'Rh', 'Pd', 'Ag', 'Cd',
                     'Hf', 'Ta', 'W', 'Re', 'Os', 'Ir', 'Pt', 'Au', 'Hg',
                     'Pb', 'Bi','Sn', 'Sb']

# Define the path to save the output files
output_dir = 'felix'
os.makedirs(output_dir, exist_ok=True)

# Loop through each transition metal and create a modified CIF file
for metal in transition_metals:
    # Create the bcc (111) slab of iron
    original_atoms = bcc111(symbol= "Fe", size=(3, 3, 3), a=2.86, vacuum=10, orthogonal=False)

    # Make a copy of the original atoms
    atoms = original_atoms.copy()

    # Replace a specific atom (e.g., the 5th atom) with the transition metal
    atoms[22].symbol = metal

    # Write the modified structure to a new CIF file
    output_filename = os.path.join(output_dir, f'Fe_dopedwith_{metal}.cif')
    write(output_filename, atoms)

print("Fe atoms have been successfully been replaced with each transition metal and saved to separate CIF files.")
```

```python
import os
import pandas as pd
from ase.io import read
from matminer.featurizers.structure import CoulombMatrix, BagofBonds
from pymatgen.io.ase import AseAtomsAdaptor

# Directory containing the CIF files
cif_dir = "felix"

# Initialize Matminer featurizers
coulomb_matrix = CoulombMatrix(flatten=True)  # Optionally flatten to turn matrix into a feature vector
bag_of_bonds = BagofBonds()       # Flatten to get a single feature vector

# List to hold data
feature_data = []

# Loop through CIF files and featurize each structure
for cif_file in os.listdir(cif_dir):
    if cif_file.endswith(".cif"):
        # Load structure using ASE and convert to pymatgen Structure
        structure = read(os.path.join(cif_dir, cif_file))
        pmg_structure = AseAtomsAdaptor.get_structure(structure)

        # Fit the Coulomb Matrix to the current structure and then featurize
        coulomb_matrix.fit(pmg_structure)
        cm_features = coulomb_matrix.featurize(pmg_structure)

        # Featurize using Bag of Bonds
        bob_features = bag_of_bonds.featurize(pmg_structure)

        # Combine all features
        combined_features = cm_features + bob_features

        # Store the result along with the CIF filename
        feature_data.append([cif_file] + combined_features)

# Get feature labels (CoulombMatrix + BagofBonds)
cm_labels = coulomb_matrix.feature_labels()
bob_labels = bag_of_bonds.feature_labels()

# Create a DataFrame with appropriate feature labels
columns = ['cif_filename'] + cm_labels + bob_labels
df = pd.DataFrame(feature_data, columns=columns)

# Save to CSV
output_csv = "doped_iron_slab_features_bob_cm.csv"
df.to_csv(output_csv, index=False)

print(f"Features saved to {output_csv}")
```

```python
import os
import pandas as pd
from ase.io import read
from matminer.featurizers.structure import CoulombMatrix, BagofBonds

# Directory containing the CIF files
cif_dir = "felix"

# Initialize Matminer featurizers
coulomb_matrix = CoulombMatrix(flatten=True)  # Optionally flatten to turn matrix into a feature vector
bag_of_bonds = BagofBonds()        # Flatten to get a single feature vector

# List to hold data
feature_data = []

# Loop through CIF files and featurize each structure
for cif_file in os.listdir(cif_dir):
    if cif_file.endswith(".cif"):
        # Load structure using ASE
        structure = read(os.path.join(cif_dir, cif_file))

        # Check if the structure has valid attributes before proceeding
        if structure is not None:
            # Fit the Coulomb Matrix and Bag of Bonds to the current structure
            coulomb_matrix.fit(structure)
            cm_features = coulomb_matrix.featurize(structure)

            bag_of_bonds.fit(structure)
            bob_features = bag_of_bonds.featurize(structure)

            # Combine all features
            combined_features = cm_features + bob_features

            # Store the result along with the CIF filename
            feature_data.append([cif_file] + combined_features)
        else:
            print(f"Warning: The structure for {cif_file} is not valid or empty.")

# Get feature labels (CoulombMatrix + BagofBonds)
cm_labels = coulomb_matrix.feature_labels()
bob_labels = bag_of_bonds.feature_labels()

# Create a DataFrame with appropriate feature labels
columns = ['cif_filename'] + cm_labels + bob_labels
df = pd.DataFrame(feature_data, columns=columns)

# Save to CSV
output_csv = "doped_iron_slab_features_bob_cm.csv"
df.to_csv(output_csv, index=False)

print(f"Features saved to {output_csv}")
```
